```python
import pandas as pd

data = {
    'Weight':[303,370,298,277,377,299,382,374,303,309,359,366,311,302,373,305,371],
    'Colour':[3,1,3,3,4,3,1,4,4,3,1,1,3,3,4,3,3],
    'Label':['Banana','Apple','Banana','Banana','Apple','Banana','Apple',
             'Apple','Banana','Banana','Apple','Apple','Banana',
             'Banana','Apple','Banana','Apple']
}

df = pd.DataFrame(data)

df.head()
```

```python
df['Fruit_Size'] = df['Weight'].apply(
    lambda x: 'Heavy' if x > 340 else 'Light'
)

df.head()
```

```python
df['Label_Code'] = df['Label'].map({
    'Apple':1,
    'Banana':0
})

df.head()
```
